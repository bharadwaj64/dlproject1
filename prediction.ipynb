```python
import tensorflow as tf 
from tensorflow.keras.models import load_model 
import pickle 
import pandas as pd 
import numpy as np
```

```python
model = load_model('model.h5')
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder = pickle.load(file)

with open('onehot_encoder_gender.pkl','rb') as file:
    onehot_encoder = pickle.load(file)

with open('scalar.pkl','rb') as file:
    scalar = pickle.load(file)
```

```python
input_data['Gender'] = label_encoder.transform([input_data['Gender']])
```

```python
df = pd.DataFrame(input_data)
```

```python
df
```

```python
data = onehot_encoder.transform([[input_data['Geography']]])
df = pd.concat([df,pd.DataFrame(data,columns=onehot_encoder.get_feature_names_out())],axis=1)
```

```python
df.drop('Geography',axis=1,inplace=True)
```

```python
df
```

```python
df_scaled = scalar.transform(df)
```

```python
predicted_value = model.predict(df_scaled)
print(predicted_value[0])
```
